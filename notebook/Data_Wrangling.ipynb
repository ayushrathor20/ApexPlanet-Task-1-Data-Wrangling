```python
numerical_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ).sum()

    print(f"\nColumn: {col}")
    print(f"Lower Limit: {lower_limit}")
    print(f"Upper Limit: {upper_limit}")
    print(f"Outliers: {outlier_count}")
```

DATA CLEANING SUMMARY

1. Initial Dataset:
   - Rows: 1000
   - Columns: 12

2. Missing Values:
   - Age: 20 missing values
   - Treatment: Filled using median value (41.0)
   - City: 13 missing values
   - Treatment: Filled with "Unknown"

3. Duplicate Records:
   - Duplicate rows found: 0
   - No duplicate records were removed.

4. Data Formatting:
   - Text fields were standardized by removing extra spaces.
   - Categorical fields were standardized for consistent formatting.
   - Order_Date was converted to datetime format.

5. Sales Validation:
   - Total_Sales was validated against Quantity × Unit_Price.
   - Actual mismatches found: 0 after rounding to 2 decimal places.

6. Outlier Analysis:
   - Age: 0 outliers
   - Quantity: 0 outliers
   - Unit_Price: 0 outliers
   - Total_Sales: 19 outliers
   - The 19 Total_Sales outliers were retained because they may represent legitimate high-value transactions.

7. Feature Engineering:
   - Created Year from Order_Date.
   - Created Month from Order_Date.
   - Created Month_Name from Order_Date.

8. Final Dataset:
   - Rows: 1000
   - Columns: 15
   - Dataset is ready for Exploratory Data Analysis.

```python
script = '''
import pandas as pd

# 1. Load Dataset
df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")

# 2. Handle Missing Values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["City"] = df["City"].fillna("Unknown")

# 3. Remove Duplicate Records
df = df.drop_duplicates()

# 4. Clean Text Columns
text_columns = [
    "Order_ID",
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "City",
    "Product",
    "Category"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

# 5. Standardize Categorical Columns
df["Gender"] = df["Gender"].str.title()
df["City"] = df["City"].str.title()
df["Category"] = df["Category"].str.title()

# 6. Convert Order_Date to Datetime
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

# 7. Feature Engineering
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.month_name()

# 8. Validate Sales Calculation
df["Calculated_Sales"] = (
    df["Quantity"] * df["Unit_Price"]
).round(2)

df["Sales_Difference"] = (
    df["Total_Sales"] - df["Calculated_Sales"]
).round(2)

print(
    "Sales Calculation Mismatches:",
    (df["Sales_Difference"] != 0).sum()
)

# 9. Remove Temporary Validation Columns
df.drop(
    columns=[
        "Calculated_Sales",
        "Sales_Difference"
    ],
    inplace=True
)

# 10. Final Validation
print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Rows:", df.duplicated().sum())

# 11. Save Cleaned Dataset
df.to_excel(
    "Cleaned_ApexPlanet_Sales_Dataset.xlsx",
    index=False
)

print("Data cleaning completed successfully!")
print("Final Dataset Shape:", df.shape)
'''

with open("data_cleaning.py", "w") as file:
    file.write(script)

print("Python script created successfully!")
```
